```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import simpson

x = np.linspace(0,1,100)
y = np.exp(-x/0.99)

start_index = 15
end_index = -15

s_y = simpson(y[start_index:end_index],x=x[start_index:end_index])


x0 = x[start_index]
x1 = x[end_index]
y0 = y[start_index]
y1 = y[end_index]

m = (y1-y0)/(x1-x0)

y_linear = y0 + m*(x - x0)

y_equiv = m*(x - x0) + s_y/(x1-x0) - 0.5*m*(x1-x0)

s_y_equiv = simpson(y_equiv[start_index:end_index],x=x[start_index:end_index])
print(f"s_y = {s_y}, s_y_equiv = {s_y_equiv}")

fig,ax = plt.subplots()
ax.plot(x,y,'.-',alpha = 0.4)
ax.plot(x,y_linear,'.-',alpha = 0.4)
ax.plot(x,y_equiv, '.-',alpha = .6)
plt.show()
```
