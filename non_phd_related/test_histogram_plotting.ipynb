```python
import numpy as np
import matplotlib.pyplot as plt

# create dataset
counts = np.random.poisson(20, 1000)
bins = range(100) 

fig,ax = plt.subplots()
ax.hist(counts, bins)
plt.show()
```

```python
# premake bins
binned_counts, edges = np.histogram(counts, bins=bins)

fig,ax = plt.subplots()
ax.stairs(binned_counts, bins, fill=True, alpha = 0.2)
plt.show()
```
