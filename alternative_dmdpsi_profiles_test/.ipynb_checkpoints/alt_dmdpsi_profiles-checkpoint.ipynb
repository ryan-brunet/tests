```python
import numpy as np
from matplotlib import pyplot as plt
import scipy.integrate as integrate
plt.rc('text', usetex=True)
plt.rc('font', family='serif')

# Initial magnetic field strength
B_star= 1.0e+8
# NS radius
R_star= 1.0e+4
# Solar mass
M_sol= 1.989e+30
# NS mass as a multiple of solar mass (default = 1.4)
M_NS= 1.4
# Gravitational constant
G= 6.6726e-11
# Isothermal atmosphere speed of sound
cs = 1.0e+6
m_accretion = 5.e-8
ma = m_accretion*M_sol
M_star = M_sol*M_NS
    
# Define grid size:
nr= 128
nth= 128
XMIN= 0.0
XMAX= 50000.0
MUMIN= 0.0
MUMAX= 1.0
X1LOW= 8.0
LOGSCALEX= 1
alpha_ratio = 1
dmu = np.abs(MUMAX-MUMIN)/(nth-1.0)

bratio = 10.
n_contours = nr - 1
x0 = cs*cs*R_star*R_star/(G*M_star)
psi_star = B_star*R_star**2/2 # Flux function at equator
mu0 = np.pi*4.0e-7
F0 = cs*cs*ma/x0/x0/x0
psi_0 = np.sqrt(F0)
```

```python
mu_levels = np.empty(n_contours)
for i in range(0,n_contours):
    contour_mesh_value = (n_contours - i)/(n_contours + 1.0)/dmu
    mu_levels[i] = contour_mesh_value*dmu

psi_levels = (psi_star/psi_0)*(1-mu_levels*mu_levels)/alpha_ratio
psi_levels = np.append(0,psi_levels)
```

```python
# PM04 dmdpsi
dmdpsi_pm04 = 0.5*psi_0/psi_star*bratio*np.exp(-bratio*psi_levels*psi_0/psi_star)/(1.0-np.exp(-bratio))
```

```python
# alternative profile
# dmdpsi_alt = 0.5*psi_0/psi_star*bratio*
```

```python
# integrate profile
mass = integrate.simpson(dmdpsi_pm04,psi_levels)*ma/M_sol*2
print(f"PM04 mass = {mass:0.4e}")

fig,ax = plt.subplots()
ax.plot(psi_levels*psi_0/psi_star, dmdpsi_pm04, '.-', alpha = 0.4, label='PM04')
ax.set(xlabel = r"$\psi/\psi_*$", ylabel=r"$dM/d\psi$")
plt.legend()
plt.show()
```
