```python
import numpy as np
from matplotlib import pyplot as plt
from sklearn.linear_model import LinearRegression
%matplotlib widget

ma = np.array([1.e-7, 5.e-7, 1.e-6, 5.e-6, 1.e-5])
char_lengths = np.array([8.5,126.2448,261.4128,1130.3424,1937.232])
x= np.log10(ma).reshape((-1, 1))
y = np.log10(char_lengths)
model = LinearRegression()
model.fit(x,y)
slope = model.coef_
y_int = model.intercept_

fig,ax = plt.subplots()
ax.plot(np.log10(ma),np.log10(char_lengths),'.',alpha=0.4)
ax.set(xlabel=r"$\log_{10}(M_a/M_\odot)$",ylabel=r"$\log_{10}(\ell)$")
ax.plot(np.log10(ma),slope*np.log10(ma) + y_int,alpha = 0.2)
plt.show()

   #   │ 1e-07 16.1 8.5
   # 2   │ 5e-07 239.1 126.2448
   # 3   │ 1e-06 495.1 261.4128
   # 4   │ 5e-06 2140.8 1130.3424
   # 5   │ 1e-05 3669.0 1937.232
```
