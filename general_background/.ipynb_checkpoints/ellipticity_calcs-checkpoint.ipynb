```python
import numpy as np
from matplotlib import pyplot as plt


# theoretical prediction (Melatos & Payne, 2005)
M_sol = 1.989e30
M_star = 1.4*M_sol
R_star = 1e4
G = 6.6726e-11
B_star = 1e8
psi_star = B_star*R_star**2/2
cs= 1e6
b = 3
mu0 = 4*np.pi*1.e-7
x0 = cs**2*R_star**2/G/M_star

# ellipticity data for b = 3
masses = np.array([1.e-7, 5.e-7, 1.e-6, 5.e-6, 1.e-5])
ellipticity = np.array([1.5100e-07, 7.5102e-07, 1.4875e-06, 6.9720e-06, 1.2842e-05])
ellipticity_cfd = np.array([])
theoretical_masses = np.logspace(-10,0.,100, base=10)
M_c = 2*np.pi*G*M_star*psi_star**2/cs**4/R_star**2/mu0
print(f"M_c/M_* = {M_c/M_star:0.5e}")
eps = (2.5*theoretical_masses)*(1-3/2/b)/(1+theoretical_masses*M_star*b**2/8/M_c)

fig,ax = plt.subplots()
ax.plot(masses,ellipticity, '.',alpha = 0.5)
ax.plot(theoretical_masses, eps, 'r',alpha = 0.2)
ax.set_xscale('log')
ax.set_yscale('log')
ax.set(xlim=[1.e-10,1], ylim=[1.e-10,1.e-4], xlabel=r"$M_a/M_\odot$",ylabel=r"$\epsilon$")
plt.show()
```

```python
# check simpson double integral is working as expected. Should equal 2.5
from scipy.integrate import simpson
x,dx = np.linspace(0,1,100,retstep=True)
y,dy = np.linspace(0,5,100,retstep=True)

xgrid,ygrid = np.meshgrid(x,y)
zgrid = xgrid**4*ygrid
yslice = np.array([simpson(zgrid[:,i],dx=dy) for i in range(100)])
integral = simpson(yslice,dx=dx)
print(integral)
```
