```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import RegularGridInterpolator
```

```python
def F(u, v):
    return u * np.cos(u * v) + v * np.sin(u * v)
```

```python
fit_points = [np.linspace(0, 3, 8), np.linspace(0, 3, 11)]
values = F(*np.meshgrid(*fit_points, indexing='ij'))
```

```python
print(fit_points[0])
print(fit_points[1])
```

```python
print(values)
```

```python
ut, vt = np.meshgrid(np.linspace(0, 3, 80), np.linspace(0, 3, 80), indexing='ij')
true_values = F(ut, vt)
test_points = np.array([ut.ravel(), vt.ravel()]).T

ufit,vfit = np.meshgrid(fit_points[0], fit_points[1], indexing='ij')
fitpoints = np.array([ufit.ravel(),vfit.ravel()]).T
```

```python
print(np.shape(test_points))
print(np.shape(fitpoints))
```

```python
fig,ax = plt.subplots()
ax.contour(ut,vt,true_values, levels=100, alpha = 0.3)
ax.scatter(test_points[:,0],test_points[:,1], s =2, alpha = 0.2)
ax.scatter(fitpoints[:,0],fitpoints[:,1], c='r', s = 2, alpha = 0.6)
plt.show()
```
