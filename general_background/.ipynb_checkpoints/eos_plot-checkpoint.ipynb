```python
# NS equations of state from Priymak (2011), showing the isothermal eos, the non-relativistic degenerate electrons, 
# the relativistic degenerate electrons, and the non-relativistic degenerate neutrons. Shaded regions are degenerate 
# non-relativistic electron gas (blue), degen rel electron gas (orange), degenerate non-rel neutron gas (green)
import numpy as np
from matplotlib import pyplot as plt

density = np.linspace(0,1e16,1000)
gamma_rel = 4/3
gamma_non_rel = 5/3
cs = 1e8
K_rel = 4.9e14
K_non_rel = 3.2e12
K_non_rel_neutrons = 5.4e9
pressure_rel = K_rel*density**gamma_rel
pressure_non_rel = K_non_rel*density**gamma_non_rel
pressure_iso = cs*cs*density
pressure_non_rel_neutrons = K_non_rel_neutrons*density**gamma_non_rel

fig,ax = plt.subplots()
ax.plot(density, pressure_iso, '-', label="Isothermal (A)")
ax.plot(density, pressure_non_rel, '--', label="Non-relativistic (B)")
ax.plot(density, pressure_rel, '--', label="Relativistic e- (C)")
ax.plot(density, pressure_non_rel_neutrons, '--', label="Non-relativistic neutrons (D)")
ax.axvspan(1e5,1e7, alpha = 0.2)
ax.axvspan(1e7,1e12,alpha=0.2, color='orange')
ax.axvspan(1e12,1e16,alpha=0.2, color='g')
ax.set(xscale='log',yscale='log',xlim=[1.,1.e16],ylim=[1,1e40],xlabel=r'$\rho~(\mathrm{g~cm^{-3}})$', ylabel=r"$P~(\mathrm{dyne~cm^{-2}})$")
ax.legend()
plt.grid(linestyle='--')
plt.show()
```
