```python
import numpy as np
from matplotlib import pyplot as plt

M_sol = 1.989e30
M = 1.4*M_sol

R_star = 1.e4
G = 6.67e-11

Phi_0 = G*M/R_star

r = np.linspace(0.001, 5, 100)*R_star

Phi_actual = np.zeros_like(r)

Phi_actual[r<R_star] = -G*M/2/R_star*(3 - (r[r<R_star]/R_star)**2)
Phi_actual[r>R_star] = -G*M/r[r>R_star]

Phi_transformed = Phi_actual + 2*Phi_0
Phi_approx = G*M*r/R_star**2
diff = np.abs(Phi_transformed - Phi_approx)/Phi_transformed * 100

diff_08R = diff[np.where(r>0.8*R_star)[0][0]]
print(f"pct diff @ 0.8R_star = {diff_08R:0.4f}%")

fig,ax = plt.subplots()
# ax.plot(r/R_star,Phi_actual, '-', label='actual')
ax.plot(r/R_star,Phi_transformed/Phi_0, '-', alpha = 0.7, label='actual')
ax.plot(r/R_star, Phi_approx/Phi_0, '-', alpha = 0.2, label='approx')
ax.axvline(1, c='k', alpha =0.3, linestyle='--', label=r'$R_\odot$')
ax.axvline(0.9, c='b', alpha =0.3, linestyle='--', label=r'$0.9R_\odot$')
ax.axvline(0.75, c='r', alpha =0.3, linestyle='--', label=r'$0.75R_\odot$')
ax.legend()
ax.set(xlabel = 'r/R', ylabel=r'$\Phi(r)/\Phi_0$', xlim=[0,2], ylim=[0,2])

fig,ax = plt.subplots()
ax.plot(r/R_star, diff)
ax.axvline(1, c='k', alpha =0.3, linestyle='--', label=r'$R_\odot$')
ax.axvline(0.9, c='b', alpha =0.3, linestyle='--', label=r'$0.9R_\odot$')
ax.axvline(0.75, c='r', alpha =0.3, linestyle='--', label=r'$0.75R_\odot$')
ax.set(xlabel = 'r/R', ylabel='% difference actual vs approximation', xlim=[0,2])
ax.axhline(0,c='k', alpha =0.3, linestyle='--')
ax.legend()
plt.show()
```
