```python
import numpy as np
from matplotlib import pyplot as plt
x = np.linspace(0, 1, 100)
s_b = x*2
s_adj_prev = x*3

k1 = np.zeros_like(s_b)
for kk in range(len(s_b)):
    if (s_b[kk] > s_adj_prev[kk]):
        s_big = s_b
        s_small = s_adj_prev
        temp = np.where(s_small > s_big[kk])
    else:
        s_big = s_adj_prev
        s_small = s_b
        temp = np.where(s_big > s_small[kk])
    # if (kk == 10):
    #     print(temp[0])
    if (len(temp[0]) > 0):
        k1[kk] = temp[0][0]
    else:
        k1[kk] = len(s_small)-1
k1 = k1.astype(int)
print(k1)


%matplotlib widget
fig, ax = plt.subplots()
ax.plot(x, s_b, '.-', label='s_b')
# ax.plot(x, s_adj_prev, '.-', label='s_adj_prev')
ax.scatter(x[k1], s_adj_prev[k1], alpha=0.2, s=10)
ax.legend()
plt.show()
```
