```python
def count_up(n):
    i = 0
    while i < n:
        yield i
        i += 1

for x in count_up(5):
    print(x)
```

```python
def count_up_2(n):
    i = 0
    while i < n:
        return i
        i += 1

for x in count_up_2(5):
    print(x)
```
