```python
import numpy as np
from matplotlib import pyplot as plt

n_contours=100
psi = np.linspace(0,1,n_contours)
b = 10

dmdpsi = 0.5*np.exp(-psi*b)/(1. - np.exp(-b))

fig,ax = plt.subplots()
ax.plot(psi, dmdpsi, '.')
plt.show()
```

```python
# clump masses
dmdpsi_clumped = np.zeros_like(dmdpsi)
n_clump = 5
for k in range(0,n_contours,n_clump):
    clump_mass = np.sum(dmdpsi[k:k+n_clump])
    #print(dmdpsi[k:k+n_clump], len(dmdpsi[k:k+n_clump]), clump_mass)
    dmdpsi_clumped[k:k+n_clump] = clump_mass/n_clump
    
fig,ax = plt.subplots()
ax.plot(psi, dmdpsi_clumped, '.')
ax.plot(psi, dmdpsi, '.-', alpha = 0.2)
plt.show()
```

```python
dmdpsi = np.copy(dmdpsi_clumped)
# clump masses
dmdpsi_clumped = np.zeros_like(dmdpsi)
n_clump = 5
for k in range(0,n_contours,n_clump):
    clump_mass = np.sum(dmdpsi[k:k+n_clump])
    #print(dmdpsi[k:k+n_clump], len(dmdpsi[k:k+n_clump]), clump_mass)
    dmdpsi_clumped[k:k+n_clump] = clump_mass/n_clump
    
fig,ax = plt.subplots()
ax.plot(psi, dmdpsi_clumped, '.')
ax.plot(psi, dmdpsi, '.-', alpha = 0.2)
plt.show()
```
