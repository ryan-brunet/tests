```python
# testing interpolation kernel
import numpy as np
import matplotlib.pyplot as plt

K1 = 15
K2 = 12

lmb = 0.5

x = np.linspace(0,10,1000)
x_interface = 4

y1 = K1 * np.ones_like(x)
y2 = K2 * np.ones_like(x)

w =  1-np.exp(-np.abs(x - x_interface)/lmb)
w_signed = (1 + np.where((x - x_interface) < 0, -w,w))/2
y_interpolate = ((1.0 - w_signed) * y1 + w_signed * y2)


fig,ax = plt.subplots()
ax.plot(x,y1, '--', label='y1')
ax.plot(x,y2, '--', label='y2')
ax.plot(x,y_interpolate, '-', label='yint')
ax.axvline(x_interface,linestyle='--', color='grey', alpha = 0.3)
ax.legend()
plt.show()
```
