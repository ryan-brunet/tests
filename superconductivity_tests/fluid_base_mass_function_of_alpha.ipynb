```python
import numpy as np
from matplotlib import pyplot as plt

alpha = np.linspace(0,1,100)
m = (np.pi - np.sin(np.pi*alpha) + np.pi*alpha*np.cos(np.pi*alpha))/np.pi
def m_func(alpha):
    return (np.pi - np.sin(np.pi*alpha) + np.pi*alpha*np.cos(np.pi*alpha))/np.pi
fig,ax = plt.subplots()
ax.plot(alpha, m_func(alpha), '-')
ax.set(xlabel=r"$\alpha=R_{\rm min}\,/\,R_\ast$", ylabel=r"$M_{\rm base}\,/\,M_\ast$")
plt.show()

print(f"Mb/M* = {m_func(0.998):0.4e}")
```
