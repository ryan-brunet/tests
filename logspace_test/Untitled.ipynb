```python
import numpy as np
from matplotlib import pyplot as plt

xmin = 0.1
xmax = 10
npoints = 11

# logspace places points at 10^i for i in (xmin, xmax). So if we set i_0 = log(xmin), 
# then x_0 = 10^{log(xmin)} = xmin
xmin_log = np.log10(xmin) 
xmax_log = np.log10(xmax)

# generate points
x_lin = np.linspace(xmin,xmax, npoints)
x_log = np.logspace(xmin_log, xmax_log, npoints)
y = np.ones_like(x_lin)


print(x_log)
fig, ax = plt.subplots()
ax.plot(x_lin, y, '.', label="Linearly spaced points")
ax.plot(x_log, y, '.', label="Log spaced points")
ax.legend()
plt.show()
```

```python
# now want to place points near xmax, rather than xmin:
# in the above, the set of points is determined by
# x = 10**(xmin_log - i*DX), where i = (0, N-1), DX = (xmax_log - xmin_log)/(N-1), so
# x_0 = 10**(-1.0 - 0*(2/10)) = 0.1, x_1 = 10**(-1.0* (1)*(0.2)) = 10**(-0.8) ...

x2 = xmax - x_log
print(x2)
x3= np.sort(x2)
print(x3)

fig,ax = plt.subplots()
ax.plot(x_lin, y, '.')
ax.plot(x2, y, '.')
plt.show()
```
