```python
import numpy as np
from matplotlib import pyplot as plt
%matplotlib widget

r = np.linspace(0.01,1,100)
th = np.linspace(0,np.pi/2,100)
thg,rg = np.meshgrid(th,r)
print(rg[:10,0])

fg = rg*np.cos(thg)

dfdr,dfdth = np.gradient(fg,r,th)


fig,ax = plt.subplots()
ax.contour(thg,rg,fg,levels=100)
ax.quiver(thg,rg,1/rg*dfdth,dfdr,angles='xy',scale_units='xy',scale=100)
plt.show()
```
