```python
import numpy as np
from matplotlib import pyplot as plt

x = np.linspace(0,1,100)
y = np.linspace(0,1,100)

a = 4
b = 2

def f_func(x,y,a,b):
    return a*x + b*y

xgrid,ygrid = np.meshgrid(x,y)

f = f_func(xgrid,ygrid,a,b)

c0 = 0.4

th = np.linspace(0,np.pi/2,50)

xc = c0*np.sin(th)
yc = c0*np.cos(th)

fig,ax = plt.subplots()
ax.contour(xgrid,ygrid,f,levels=50)
ax.scatter(xc,yc,s=2,alpha = 0.5)
ax.quiver(xc,yc,xc/np.sqrt(xc**2 + yc**2),yc/np.sqrt(xc**2 + yc**2),angles='xy',scale_units='xy',alpha=0.3)
ax.set_aspect('equal')
plt.show()
```
