## Another test of gradient vector
Testing what the gradient vector looks like in polar and cartesian coordinates for the function $f(r,\theta) = r^2 + \theta^2$.

```python
import numpy as np
from matplotlib import pyplot as plt

r = np.linspace(0.1,np.pi,100)
th = np.linspace(0.1,np.pi,100)

thgrid,rgrid=np.meshgrid(th,r)

xgrid = rgrid*np.cos(thgrid)
ygrid = rgrid*np.sin(thgrid)

def f(r,theta):
    return r**2 + theta**2

f = f(rgrid,thgrid)

dfdr,dfdth = np.gradient(f,r, th,edge_order=2)
# dfdth = np.gradient(f,thgrid,edge_order=2)

th_points = np.linspace(0,3,100)
r_points = np.sqrt(9 - th_points**2)

dfdr = 2*r_points
dfdth = 2*th_points

%matplotlib widget
fig,ax = plt.subplots()
# ax.contourf(xgrid,ygrid,f,levels=100)
ax.contour(thgrid,rgrid,f,levels=100)
ax.scatter(th_points,r_points,s=5)
ax.quiver(th_points,r_points,dfdth,dfdr,scale_units='xy',angles='xy',alpha = 0.2)
ax.set_aspect('equal')
plt.show()


```

```python
# now plot in Cartesian
x_points = r_points * np.cos(th_points)
y_points = r_points * np.sin(th_points)

dfdr = 2*np.sqrt(x_points**2 + y_points**2)
dfdth = 2*np.arctan(y_points/x_points)/np.sqrt(x_points**2 + y_points**2)

dfdx = x_points/np.sqrt(x_points**2 + y_points**2)*dfdr + y_points/(x_points**2 + y_points**2) * dfdth
dfdy = y_points/np.sqrt(x_points**2 + y_points**2)*dfdr - x_points/(x_points**2 + y_points**2) * dfdth

fig,ax = plt.subplots()
ax.contour(xgrid,ygrid,f,levels=100)
ax.scatter(x_points,y_points,s=5)
ax.quiver(x_points,y_points,dfdx,dfdy,angles='xy',scale_units='xy',alpha=0.2)
plt.show()
```
